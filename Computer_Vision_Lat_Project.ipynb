# Buat Meta data

```python
import os
import re
import pandas as pd

# ============================================================
# PATH DATASET
# ============================================================

# Sesuaikan dengan lokasi dataset kamu di Google Drive
dataset_path = "/content/drive/MyDrive/Dataset_Project_Comvis"

# Lokasi untuk menyimpan metadata
metadata_path = "/content/drive/MyDrive/metadata.csv"


# ============================================================
# POLA NAMA FILE
# ============================================================

# Contoh:
# original_52_1.png
# original_33_21.png
# forgeries_55_1.png
# forgeries_43_24.png

filename_pattern = re.compile(
    r"^(original|forgeries)_(\d+)_(\d+)\.png$",
    re.IGNORECASE
)


# ============================================================
# MEMBACA DATASET
# ============================================================

metadata = []
invalid_files = []

for folder in sorted(os.listdir(dataset_path), key=lambda x: int(x) if x.isdigit() else x):

    folder_path = os.path.join(dataset_path, folder)

    # Pastikan yang dibaca adalah folder
    if not os.path.isdir(folder_path):
        continue

    for filename in sorted(os.listdir(folder_path)):

        file_path = os.path.join(folder_path, filename)

        # Hanya membaca file PNG
        if not filename.lower().endswith(".png"):
            continue

        # Cek pola nama file
        match = filename_pattern.match(filename)

        if match is None:
            invalid_files.append({
                "folder": folder,
                "filename": filename,
                "reason": "Format nama file tidak sesuai"
            })
            continue

        signature_type = match.group(1).lower()
        writer_id = int(match.group(2))
        image_number = int(match.group(3))

        # Menentukan label
        if signature_type == "original":
            label = 1
            label_name = "genuine"
        else:
            label = 0
            label_name = "forged"

        metadata.append({
            "filepath": file_path,
            "filename": filename,
            "folder": folder,
            "writer_id": writer_id,
            "image_number": image_number,
            "signature_type": signature_type,
            "label": label,
            "label_name": label_name
        })


# ============================================================
# MEMBUAT DATAFRAME
# ============================================================

metadata_df = pd.DataFrame(metadata)

print("Metadata berhasil dibuat.")
print()
print("Jumlah data:", len(metadata_df))
print("Jumlah genuine:", (metadata_df["label_name"] == "genuine").sum())
print("Jumlah forged :", (metadata_df["label_name"] == "forged").sum())
```

# Membuat pasangan Siamese

```python
def create_pairs(
    df,
    pairs_per_writer=100,
    seed=42
):
    """
    Membuat balanced Siamese pairs.

    pair_label:
        0 = similar  (genuine-genuine)
        1 = dissimilar (genuine-forgery)
    """

    rng = random.Random(seed)

    pairs = []

    for writer_id in sorted(df["writer_id"].unique()):

        writer_data = df[
            df["writer_id"] == writer_id
        ]

        genuine = writer_data[
            writer_data["label_name"] == "genuine"
        ]

        forged = writer_data[
            writer_data["label_name"] == "forged"
        ]

        genuine_paths = genuine["filepath"].tolist()
        forged_paths = forged["filepath"].tolist()

        # ----------------------------------------------------
        # Similar pairs: genuine + genuine
        # ----------------------------------------------------

        possible_positive = []

        for i in range(len(genuine_paths)):
            for j in range(i + 1, len(genuine_paths)):

                possible_positive.append(
                    (
                        genuine_paths[i],
                        genuine_paths[j]
                    )
                )

        # ----------------------------------------------------
        # Dissimilar pairs: genuine + forged
        # ----------------------------------------------------

        possible_negative = []

        for genuine_path in genuine_paths:
            for forged_path in forged_paths:

                possible_negative.append(
                    (
                        genuine_path,
                        forged_path
                    )
                )

        # ----------------------------------------------------
        # Sampling
        # ----------------------------------------------------

        positive_n = min(
            pairs_per_writer,
            len(possible_positive)
        )

        negative_n = min(
            pairs_per_writer,
            len(possible_negative)
        )

        positive_pairs = rng.sample(
            possible_positive,
            positive_n
        )

        negative_pairs = rng.sample(
            possible_negative,
            negative_n
        )

        # Similar = 0
        for img1, img2 in positive_pairs:

            pairs.append({
                "image1": img1,
                "image2": img2,
                "pair_label": 0
            })

        # Dissimilar = 1
        for img1, img2 in negative_pairs:

            pairs.append({
                "image1": img1,
                "image2": img2,
                "pair_label": 1
            })

    pairs_df = pd.DataFrame(pairs)

    pairs_df = pairs_df.sample(
        frac=1,
        random_state=seed
    ).reset_index(drop=True)

    return pairs_df
```

# Fungsi membaca dan preprocessing gambar

```python
def load_image(path):

    image = tf.io.read_file(path)

    image = tf.image.decode_png(
        image,
        channels=1
    )

    image = tf.image.resize(
        image,
        [IMG_HEIGHT, IMG_WIDTH]
    )

    image = tf.cast(
        image,
        tf.float32
    ) / 255.0

    return image
```

# Membuat TensorFlow Dataset

```python
def make_tf_dataset(
    pairs_df,
    batch_size=32,
    shuffle=True
):

    image1_paths = pairs_df["image1"].values
    image2_paths = pairs_df["image2"].values

    labels = pairs_df["pair_label"].values.astype(
        np.float32
    )

    dataset = tf.data.Dataset.from_tensor_slices(
        (
            {
                "image1": image1_paths,
                "image2": image2_paths
            },
            labels
        )
    )

    if shuffle:
        dataset = dataset.shuffle(
            buffer_size=len(pairs_df),
            seed=SEED
        )

    def preprocess(inputs, label):

        image1 = load_image(
            inputs["image1"]
        )

        image2 = load_image(
            inputs["image2"]
        )

        return (
            {
                "image1": image1,
                "image2": image2
            },
            label
        )

    dataset = dataset.map(
        preprocess,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    dataset = dataset.batch(
        batch_size
    )

    dataset = dataset.prefetch(
        tf.data.AUTOTUNE
    )

    return dataset
```

# Buat CNN Feature Extractor

```python
from tensorflow.keras import layers
from tensorflow.keras import Model


def build_embedding_network():

    inputs = layers.Input(
        shape=(
            IMG_HEIGHT,
            IMG_WIDTH,
            CHANNELS
        )
    )

    x = layers.Conv2D(
        32,
        (3, 3),
        activation="relu",
        padding="same"
    )(inputs)

    x = layers.MaxPooling2D(
        (2, 2)
    )(x)

    x = layers.Conv2D(
        64,
        (3, 3),
        activation="relu",
        padding="same"
    )(x)

    x = layers.MaxPooling2D(
        (2, 2)
    )(x)

    x = layers.Conv2D(
        128,
        (3, 3),
        activation="relu",
        padding="same"
    )(x)

    x = layers.MaxPooling2D(
        (2, 2)
    )(x)

    x = layers.Conv2D(
        256,
        (3, 3),
        activation="relu",
        padding="same"
    )(x)

    x = layers.GlobalAveragePooling2D()(x)

    x = layers.Dense(
        128,
        activation="relu"
    )(x)

    x = layers.Dropout(
        0.3
    )(x)

    # Embedding akhir
    embeddings = layers.Dense(
        64,
        activation=None,
        name="embedding"
    )(x)

    return Model(
        inputs,
        embeddings,
        name="signature_embedding_network"
    )
```

```python
import tensorflow as tf

# Cache dalam tensor TensorFlow
image_tensor = tf.constant(images, dtype=tf.float32)

def make_fast_dataset(i1, i2, labels, batch_size=32, shuffle=False):
    ds = tf.data.Dataset.from_tensor_slices((i1, i2, labels))

    if shuffle:
        ds = ds.shuffle(
            buffer_size=len(labels),
            seed=42
        )

    def fetch_images(idx1, idx2, label):
        img1 = tf.gather(image_tensor, idx1)
        img2 = tf.gather(image_tensor, idx2)

        return (
            {"image1": img1, "image2": img2},
            label
        )

    ds = ds.map(
        fetch_images,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    ds = ds.batch(batch_size)
    ds = ds.prefetch(tf.data.AUTOTUNE)

    return ds

BATCH_SIZE = 32

train_dataset = make_fast_dataset(
    train_i1, train_i2, train_y,
    BATCH_SIZE, shuffle=True
)

val_dataset = make_fast_dataset(
    val_i1, val_i2, val_y,
    BATCH_SIZE, shuffle=False
)

test_dataset = make_fast_dataset(
    test_i1, test_i2, test_y,
    BATCH_SIZE, shuffle=False
)

print("Dataset siap untuk training.")
```

# Buat Siamese Network

```python
from tensorflow.keras import backend as K


def euclidean_distance(vectors):

    vector1, vector2 = vectors

    squared_difference = tf.square(
        vector1 - vector2
    )

    sum_squared_difference = tf.reduce_sum(
        squared_difference,
        axis=1,
        keepdims=True
    )

    distance = tf.sqrt(
        tf.maximum(
            sum_squared_difference,
            K.epsilon()
        )
    )

    return distance
```

```python
def build_siamese_model():

    embedding_network = build_embedding_network()

    input1 = layers.Input(
        shape=(
            IMG_HEIGHT,
            IMG_WIDTH,
            CHANNELS
        ),
        name="image1"
    )

    input2 = layers.Input(
        shape=(
            IMG_HEIGHT,
            IMG_WIDTH,
            CHANNELS
        ),
        name="image2"
    )

    embedding1 = embedding_network(
        input1
    )

    embedding2 = embedding_network(
        input2
    )

    distance = layers.Lambda(
        euclidean_distance,
        name="euclidean_distance"
    )(
        [embedding1, embedding2]
    )

    model = Model(
        inputs=[input1, input2],
        outputs=distance,
        name="Siamese_CNN"
    )

    return model
```

# Contrastive Loss

```python
def contrastive_loss(
    y_true,
    y_pred,
    margin=1.0
):

    y_true = tf.cast(
        y_true,
        tf.float32
    )

    # Similar pair:
    # y_true = 0
    # distance harus kecil

    positive_loss = (
        (1 - y_true)
        * tf.square(y_pred)
    )

    # Dissimilar pair:
    # y_true = 1
    # distance harus lebih besar dari margin

    negative_loss = (
        y_true
        * tf.square(
            tf.maximum(
                margin - y_pred,
                0
            )
        )
    )

    loss = (
        positive_loss
        + negative_loss
    )

    return tf.reduce_mean(loss)
```

# Compile model

```python
siamese_model = build_siamese_model()

siamese_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0001
    ),
    loss=contrastive_loss
)

siamese_model.summary()
```
